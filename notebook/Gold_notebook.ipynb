```python
silver_jobs_df = spark.table("silver_jobs")

silver_jobs_df.printSchema()
```

```python
from pyspark.sql.functions import length, col

silver_jobs_df = silver_jobs_df.withColumn(
    "description_length",
    length(col("description"))
)
```

```python
silver_jobs_df.select(
    "job_id",
    "job_title",
    "description_length"
).show(20, truncate=False)
```

```python
silver_jobs_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("silver_jobs")
```

```python
silver_jobs_df = spark.table("silver_jobs")

print("Silver rows:", silver_jobs_df.count())
```

```python
gold_jobs_df = silver_jobs_df.select(
    "job_id",
    "job_title",
    "company",
    "publication_date",
    "application_deadline",
    "publication_date_only",
    "publication_year",
    "publication_month",
    "days_to_deadline",
    "employment_type",
    "working_hours",
    "duration",
    "workplace_model",
    "city",
    "region",
    "occupation",
    "occupation_field",
    "description_length",
    "has_power_bi",
    "has_sql",
    "has_python",
    "has_azure",
    "has_databricks",
    "has_fabric",
    "job_url"
)
```

```python
print("Gold rows:", gold_jobs_df.count())
print("Unique job IDs:", gold_jobs_df.select("job_id").distinct().count())
```

```python
gold_jobs_df.show(5, truncate=False)
```

```python
gold_jobs_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("gold_jobs")
```

```python
gold_check_df = spark.table("gold_jobs")

print("Gold rows:", gold_check_df.count())
print("Unique job IDs:", gold_check_df.select("job_id").distinct().count())
```

```python
from pyspark.sql.functions import col

gold_skills_df = gold_jobs_df.selectExpr(
    "job_id",
    """
    stack(
        6,
        'Power BI', has_power_bi,
        'SQL', has_sql,
        'Python', has_python,
        'Azure', has_azure,
        'Databricks', has_databricks,
        'Microsoft Fabric', has_fabric
    ) as (skill, has_skill)
    """
).filter(
    col("has_skill") == True
).drop("has_skill")
```

```python
gold_skills_df.show(50, truncate=False)
```

```python
gold_skills_df.groupBy("skill") \
    .count() \
    .orderBy(col("count").desc()) \
    .show()
```

```python
gold_skills_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("gold_skills")
```

```python
gold_skills_check_df = spark.table("gold_skills")

print("Gold skills rows:", gold_skills_check_df.count())
print("Unique jobs:", gold_skills_check_df.select("job_id").distinct().count())
```
